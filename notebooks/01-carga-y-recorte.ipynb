```python
import pandas as pd

columnas = ["loan_amnt", "term", "int_rate", "grade", "sub_grade", "emp_length",
            "home_ownership", "annual_inc", "verification_status", "issue_d",
            "loan_status", "purpose", "addr_state", "dti", "delinq_2yrs",
            "fico_range_low", "revol_util", "total_pymnt", "recoveries",
            "total_rec_prncp", "total_rec_int", "total_rec_late_fee", "collection_recovery_fee"]

df = pd.read_csv("../data/accepted_2007_to_2018Q4.csv.gz", usecols=columnas, low_memory=False)
print(df.shape)
```

```python
df["issue_d"] = pd.to_datetime(df["issue_d"], format="%b-%Y")
df = df[(df["term"].str.strip() == "36 months") & df["issue_d"].between("2012-01-01", "2015-12-31")]
df = df[df["loan_status"].isin(["Fully Paid", "Charged Off"])].copy()
df["default"] = (df["loan_status"] == "Charged Off").astype(int)
```

```python
print("Filas:", len(df))
print("Tasa de default:", round(df["default"].mean() * 100, 2), "%")
print(df.isna().sum())
```

```python
df.to_csv("../data/prestamos_36m_2012_2015.csv", index=False)
```
